## 3. Limpeza e tratamento dos dados

Nesta etapa, serão removidas colunas sem informação útil, tratados valores inconsistentes, eliminadas duplicatas e ajustados os tipos de dados necessários para a análise.

```python
# Remoção das colunas completamente vazias e sem informações úteis para a análise
df = df.drop(columns=["Unnamed: 10", "Unnamed: 11", "Unnamed: 12", "Unnamed: 13"])
df.columns
```

```python
# Substituição das categorias ausentes identificadas como "#N/D"
df["PR_CAT"] = df["PR_CAT"].replace("#N/D", "Sem Categoria")
df["PR_CAT"].value_counts()
```

```python
# Armazena a quantidade de linhas antes da remoção das duplicatas
linhas_antes = len(df)

# Remove registros completamente duplicados
df = df.drop_duplicates()

# Calcula a quantidade de registros removidos
linhas_depois = len(df)
linhas_removidas = linhas_antes - linhas_depois
print("Linhas antes:", linhas_antes)
print("Linhas depois:", linhas_depois)
print("Linhas duplicadas removidas:", linhas_removidas)
```

```python
# Confirma se ainda existem registros duplicados após o tratamento
df.duplicated().sum()
```

```python
# Verificando se o Pandas está tratando data como object/texto
df["DATA"].dtype
```

```python
# Verificando o formato visual da data
df["DATA"].head()
```

```python
# Converte a coluna DATA de texto para datetime, considerando o dia na primeira posição
df["DATA"] = pd.to_datetime(df["DATA"], dayfirst=True)
df["DATA"].dtype
```

## 4. Estatística descritiva

Nesta etapa serão analisadas estatísticas descritivas da coluna CL_FHL, que representa o número de filhos dos clientes.

```python
# Verificando a quantidade de clientes que possuem 1, 2, 3 e 4 filhos / sort_index() - ordenando os números
df["CL_FHL"].value_counts().sort_index()
```

```python
# Verificando a quantidade de IDs únicos
df["CL_ID"].nunique()
```

```python
# Cria uma base com apenas um registro por cliente
clientes = df[["CL_ID", "CL_FHL"]].drop_duplicates()

# Verifica a quantidade de clientes únicos na nova base
clientes.shape
```

```python
# Calcula a média do número de filhos dos clientes
media = clientes["CL_FHL"].mean()

# Calcula a mediana do número de filhos dos clientes
mediana = clientes["CL_FHL"].median()

print("Média:", media)
print("Mediana:", mediana)
```

```python
# Calcula a moda do número de filhos dos clientes

moda = clientes["CL_FHL"].mode()
print("Moda:", moda)
```

```python
# Calcula o menor número de filhos entre os clientes

minimo = clientes["CL_FHL"].min()

# Calcula o maior número de filhos entre os clientes

maximo= clientes["CL_FHL"].max()

# Conta a quantidade de registros utilizados na análise

contagem = clientes["CL_FHL"].count()

print("Mínimo:", minimo)
print("Máximo:", maximo)
print("Contagem:", contagem)
```

```python
# Calcula o desvio padrão do número de filhos dos clientes

desvio_padrao = clientes["CL_FHL"].std()
print("Desvio padrão:", round(desvio_padrao, 2))
```

```python
# Calcular os quartis do número de filhos dos clientes

q1 = clientes["CL_FHL"].quantile(0.25)
q2 = clientes["CL_FHL"].quantile(0.50)
q3 = clientes["CL_FHL"].quantile(0.75)

print("Q1 (25%)", q1)
print("Q2 (50%)", q2)
print("Q3 (75%)", q3)
```
