## 3. Limpeza e tratamento dos dados

Nesta etapa, serão removidas colunas sem informação útil, tratados valores inconsistentes, eliminadas duplicatas e ajustados os tipos de dados necessários para a análise.

```python
# Remoção das colunas completamente vazias e sem informações úteis para a análise

df = df.drop(columns=["Unnamed: 10", "Unnamed: 11", "Unnamed: 12", "Unnamed: 13"])
df.columns
```

```python
# Substituição das categorias ausentes identificadas como "#N/D, aplicação da função com .apply e verificação da contagem com o value_counts()"

def tratar_categoria(categoria):
  if categoria == "#N/D":
    return "Sem categoria"
  else:
    return categoria
df["PR_CAT"] = df["PR_CAT"].apply(tratar_categoria)
df["PR_CAT"].value_counts()
```

```python
# Armazena a quantidade de linhas antes da remoção das duplicatas

linhas_antes = len(df)

# Remove registros completamente duplicados

df = df.drop_duplicates()

# Calcula a quantidade de registros removidos

linhas_depois = len(df)
linhas_removidas = linhas_antes - linhas_depois
print("Linhas antes:", linhas_antes)
print("Linhas depois:", linhas_depois)
print("Linhas duplicadas removidas:", linhas_removidas)
```

```python
# Confirma se ainda existem registros duplicados após o tratamento

df.duplicated().sum()
```

```python
# Verificando se o Pandas está tratando data como object/texto

df["DATA"].dtype
```

```python
# Converte a coluna DATA de texto para datetime, considerando o dia na primeira posição

df["DATA"] = pd.to_datetime(df["DATA"], dayfirst=True)
df["DATA"].dtype
```

## 4. Estatística descritiva

Nesta etapa serão analisadas estatísticas descritivas da coluna CL_FHL, que representa o número de filhos dos clientes.

```python
# Verificando a quantidade de clientes que possuem 1, 2, 3 e 4 filhos / sort_index() - ordenando os números

df["CL_FHL"].value_counts().sort_index()
```

```python
# Verificando a quantidade de IDs únicos

df["CL_ID"].nunique()
```

```python
# Cria uma base com apenas um registro por cliente

clientes = df[["CL_ID", "CL_FHL"]].drop_duplicates()

# Verifica a quantidade de clientes únicos na nova base

clientes.shape
```

```python
# Calcula a média do número de filhos dos clientes

media = clientes["CL_FHL"].mean()

# Calcula a mediana do número de filhos dos clientes

mediana = clientes["CL_FHL"].median()

print("Média:", media)
print("Mediana:", mediana)
```

```python
# Calcula a moda do número de filhos dos clientes

moda = clientes["CL_FHL"].mode()
print("Moda:", moda)
```

```python
# Calcula o menor número de filhos entre os clientes

minimo = clientes["CL_FHL"].min()

# Calcula o maior número de filhos entre os clientes

maximo= clientes["CL_FHL"].max()

# Conta a quantidade de registros utilizados na análise

contagem = clientes["CL_FHL"].count()

print("Mínimo:", minimo)
print("Máximo:", maximo)
print("Contagem:", contagem)
```

```python
# Calcula o desvio padrão do número de filhos dos clientes

desvio_padrao = clientes["CL_FHL"].std()
print("Desvio padrão:", round(desvio_padrao, 2))
```

```python
# Calcular os quartis do número de filhos dos clientes

q1 = clientes["CL_FHL"].quantile(0.25)
q2 = clientes["CL_FHL"].quantile(0.50)
q3 = clientes["CL_FHL"].quantile(0.75)

print("Q1 (25%)", q1)
print("Q2 (50%)", q2)
print("Q3 (75%)", q3)
```

## 5. Validação do identificador de compra

Nesta etapa será analisada a coluna `CO_ID`, responsável por identificar as compras realizadas. Como cada linha da base representa um item vendido, um mesmo `CO_ID` pode aparecer em vários registros quando uma compra possui mais de um produto.

A análise tem como objetivo verificar a quantidade de compras únicas e confirmar que a repetição do `CO_ID` não representa necessariamente uma duplicidade, mas diferentes itens pertencentes à mesma compra.

```python
# Verifica a quantidade de compras únicas na base

quantidade_compras = df["CO_ID"].nunique()
print("Quantidade de compras:", quantidade_compras)
```

```python
# Agrupa os registros pelo identificador da compra e conta quantos itens existem em cada uma

itens_por_compra = df.groupby("CO_ID").size()
itens_por_compra.head()
```

```python
# Visualiza os registros associados a uma compra específica

df[df["CO_ID"] == 1000].head(10)
```

### Validação do CO_ID

A análise confirmou que o `CO_ID` identifica uma compra e pode aparecer em vários registros, pois cada linha da base representa um item associado à compra. Foram identificadas 18.471 compras únicas.

Por exemplo, o `CO_ID 1000` possui 46 registros associados, contendo diferentes produtos. Portanto, a repetição do `CO_ID` não deve ser considerada uma duplicidade por si só.

## 6. Agrupamento e análise dos dados

Nesta etapa serão realizados agrupamentos para analisar os dados de diferentes perspectivas. O objetivo é resumir informações da base e identificar padrões relacionados às compras, aos clientes e aos produtos.

```python
# Agrupa os registros por gênero e conta a quantidade de compras únicas

compras_por_genero = df.groupby("CL_GENERO")["CO_ID"].nunique()
print(compras_por_genero)
```

```python
# Agrupa os registros por categoria, conta a quantidade de itens e ordena o resultado do maior para o menor

itens_por_categoria = df.groupby("PR_CAT").size().sort_values(ascending=False)

print(itens_por_categoria)
```

```python
# Agrupa os registros por gênero e compara compras e clientes únicos

compras_por_genero = df.groupby("CL_GENERO")["CO_ID"].nunique()
clientes_por_genero = df.groupby("CL_GENERO")["CL_ID"].nunique()

print("Compras por gênero:")
print(compras_por_genero)

print("\nClientes por gênero:")
print(clientes_por_genero)
```

```python
# Calcula a média de compras por cliente em cada gênero

media_compras_por_cliente = round(compras_por_genero / clientes_por_genero, 2)
print("Média de compras por cliente:")
print(media_compras_por_cliente)
```
